```python
# AIRWISE - Step 23: Time-based feature engineering

# Copy cleaned dataset
df_features = df_clean.copy()

# Convert Datetime index into a regular column
df_features = df_features.reset_index()

# Extract time-based features
df_features["Year"] = df_features["Datetime"].dt.year
df_features["Month"] = df_features["Datetime"].dt.month
df_features["Day"] = df_features["Datetime"].dt.day
df_features["Hour"] = df_features["Datetime"].dt.hour
df_features["DayOfWeek"] = df_features["Datetime"].dt.dayofweek

df_features["IsWeekend"] = (
    df_features["DayOfWeek"] >= 5
).astype(int)

# Cyclical encoding
import numpy as np

df_features["Hour_Sin"] = np.sin(
    2 * np.pi * df_features["Hour"] / 24
)

df_features["Hour_Cos"] = np.cos(
    2 * np.pi * df_features["Hour"] / 24
)

print("FEATURE ENGINEERING SUCCESSFUL")
print("=" * 45)

display(df_features.head())

print("Dataset shape:", df_features.shape)
print("Datetime column available:", "Datetime" in df_features.columns)
```

```python
import numpy as np
import pandas as pd

# Reset Datetime from the index into a normal column
df_features = df_clean.copy().reset_index()

# Sort records chronologically
df_features = df_features.sort_values("Datetime").reset_index(drop=True)

# Extract time-based features
df_features["Year"] = df_features["Datetime"].dt.year
df_features["Month"] = df_features["Datetime"].dt.month
df_features["Day"] = df_features["Datetime"].dt.day
df_features["Hour"] = df_features["Datetime"].dt.hour
df_features["DayOfWeek"] = df_features["Datetime"].dt.dayofweek

# Encode hour cyclically
df_features["Hour_Sin"] = np.sin(2 * np.pi * df_features["Hour"] / 24)
df_features["Hour_Cos"] = np.cos(2 * np.pi * df_features["Hour"] / 24)

# Create previous-hour CO measurements
df_features["CO_Lag_1"] = df_features["CO(GT)"].shift(1)
df_features["CO_Lag_2"] = df_features["CO(GT)"].shift(2)
df_features["CO_Lag_3"] = df_features["CO(GT)"].shift(3)
df_features["CO_Lag_24"] = df_features["CO(GT)"].shift(24)

# Create the next-hour CO prediction target
df_features["CO_Target_Next_Hour"] = df_features["CO(GT)"].shift(-1)

# Preview the result
display(df_features.head())
print("Feature dataset shape:", df_features.shape)
print("Datetime column available:", "Datetime" in df_features.columns)
```

```python
feature_columns = [
    "CO_Lag_1",
    "CO_Lag_2",
    "CO_Lag_3",
    "CO_Lag_24",
    "PT08.S1(CO)",
    "PT08.S2(NMHC)",
    "PT08.S3(NOx)",
    "PT08.S4(NO2)",
    "PT08.S5(O3)",
    "T",
    "RH",
    "AH",
    "Hour",
    "DayOfWeek",
    "Month",
    "Hour_Sin",
    "Hour_Cos"
]

target_column = "CO_Target_Next_Hour"

print("Number of input features:", len(feature_columns))
print("Prediction target:", target_column)
```

```python
# Keep only rows with an observed target
model_data = df_features.dropna(
    subset=[target_column]
).copy()

# Sort chronologically
model_data = model_data.sort_values("Datetime").reset_index(drop=True)

# Split at the 80% point
split_index = int(len(model_data) * 0.80)

train_data = model_data.iloc[:split_index].copy()
test_data = model_data.iloc[split_index:].copy()

print("Training records:", len(train_data))
print("Testing records:", len(test_data))

print("Training period:")
print(train_data["Datetime"].min(), "to", train_data["Datetime"].max())

print("Testing period:")
print(test_data["Datetime"].min(), "to", test_data["Datetime"].max())
```

```python
from sklearn.impute import SimpleImputer

# Separate features and target
X_train = train_data[feature_columns].copy()
X_test = test_data[feature_columns].copy()

y_train = train_data[target_column].copy()
y_test = test_data[target_column].copy()

# Learn missing-value replacements from training data only
imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

print("Training features:", X_train_imputed.shape)
print("Testing features:", X_test_imputed.shape)
print("Training targets:", y_train.shape)
print("Testing targets:", y_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_imputed, y_train)

print("Random Forest model trained successfully!")
```

```python
y_pred = rf_model.predict(X_test_imputed)

print("First 10 actual values:")
print(y_test.head(10).to_list())

print("\nFirst 10 predicted values:")
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Random Forest Performance")
print("-------------------------")
print(f"MAE:  {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²:   {r2:.4f}")
```
